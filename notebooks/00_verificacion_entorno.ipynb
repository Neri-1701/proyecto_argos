# 00 – Verificación del Entorno ARGOS

Este notebook tiene como objetivo validar la correcta configuración del entorno virtual, las dependencias y la estructura base del proyecto **ARGOS** (*Sistema Inteligente de Clasificación y Localización de Materiales*).

## 1️.- Información del entorno y dependencias

```python
import sys
import os
import platform
import pandas as pd
import numpy as np
import sklearn
import mlflow
import dvc

print('Python version:', sys.version)
print('Platform:', platform.platform())
print('Current working directory:', os.getcwd())

print('\nLibrerías instaladas:')
for pkg in [pd, np, sklearn, mlflow, dvc]:
    print(f' - {pkg.__name__:<10} {pkg.__version__}')
```

## 2️.- Verificación de la estructura del proyecto

```python
import os

# Subir un nivel desde la carpeta notebooks al root del proyecto
base_path = os.path.abspath(os.path.join(os.getcwd(), ".."))
folders = ['data', 'notebooks', 'src', 'reports', 'tests']

print('Estructura de carpetas esperada:')
for f in folders:
    path = os.path.join(base_path, f)
    status = 'OK' if os.path.exists(path) else 'FALTA'
    print(f'{f:<10} -> {status}')

print(f'\nBase path verificado:\n{base_path}')
```

## 3️.- Prueba de lectura y escritura de datos
Para comprobar el correcto funcionamiento de `pandas` y los permisos de escritura del entorno, se creará un archivo temporal en `data/raw/`.

```python
os.makedirs('data/raw', exist_ok=True)

data_test = pd.DataFrame({
    'codigo_material': ['3000001', '3000002', '3000003'],
    'descripcion': [
        'VÁLVULA ESFÉRICA 2" 150LB',
        'EMPACADURA SPW 4" 600LB',
        'ESPÁRRAGO ASTM A193 B7 1/2" X 80MM'
    ]
})

ruta_archivo = 'data/raw/test_materiales.csv'
data_test.to_csv(ruta_archivo, index=False)
print(f'Archivo de prueba generado en: {ruta_archivo}')

pd.read_csv(ruta_archivo).head()
```
